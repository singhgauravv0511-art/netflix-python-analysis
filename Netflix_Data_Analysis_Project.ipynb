# **Visualize Data**

**Plot a Bar Plot to compare Top 5 Countries with Most Movies**

```python
top_5_countries_by_movie_releases = netflix_data[netflix_data['type'] == 'Movie'].groupby('country').agg(
    Number_Of_Releases = ('show_id', 'count')
).sort_values(by='Number_Of_Releases', ascending=False).head().reset_index()
```

```python
top_5_countries_by_movie_releases
```

```python
sns.set_style('whitegrid')
plt.figure(figsize=(12, 3))
plt.title("Number of Movie Releases by Top 5 Countries")
plt.xlabel("Country")
plt.ylabel("Number of Movie Releases")
sns.barplot(data=top_5_countries_by_movie_releases, x='country', y='Number_Of_Releases', hue='country', width=.5)
plt.show()
```

**Plot a Bar Plot to compare Top 5 Countries with Most TV Shows**

```python
top_5_countries_by_tv_show_releases = netflix_data[netflix_data['type'] == 'TV Show'].groupby('country').agg(
    Number_Of_Releases = ('show_id', 'count')
).sort_values(by='Number_Of_Releases', ascending=False).head().reset_index()
```

```python
top_5_countries_by_tv_show_releases
```

```python
plt.figure(figsize=(12, 3))
plt.title("Number of TV Show Releases by Top 5 Countries")
plt.xlabel("Country")
plt.ylabel("Number of TV Show Releases")
sns.barplot(data=top_5_countries_by_tv_show_releases, x='country', y='Number_Of_Releases', hue='country', width=.5)
plt.show()
```

**Plot a Pie Chart to show the Percentages of Movies Released by Top 10 Genres**

```python
top_10_genres_by_movies_releases = netflix_data[netflix_data['type'] == 'Movie'].groupby('listed_in').agg(
    Number_Of_Releases = ('show_id', 'count')
).sort_values(by='Number_Of_Releases', ascending=False).head(10).reset_index()
```

```python
top_10_genres_by_movies_releases
```

```python
plt.title("Top 10 Genres By Number of Movies Released")
plt.pie(
    top_10_genres_by_movies_releases['Number_Of_Releases'],
    labels=top_10_genres_by_movies_releases['listed_in'],
    autopct="%.2f%%"
)
plt.show()
```

**Plot a Pie Chart to show the Percentages of TV Shows Released by Top 10 Genres**

```python
top_10_genres_by_tv_show_releases = netflix_data[netflix_data['type'] == 'TV Show'].groupby('listed_in').agg(
    Number_Of_Releases = ('show_id', 'count')
).sort_values(by='Number_Of_Releases', ascending=False).head(10).reset_index()
```

```python
top_10_genres_by_tv_show_releases
```

```python
plt.title("Top 10 Genres By Number of TV Shows Released")
plt.pie(
    top_10_genres_by_tv_show_releases['Number_Of_Releases'],
    labels=top_10_genres_by_tv_show_releases['listed_in'],
    autopct="%.2f%%"
)
plt.show()
```

**Plot a Line Plot to compare the Trend Line of Numbers of Movies and TV Shows released across all the years**

```python
number_of_releases_by_release_year = netflix_data.groupby(['release_year', 'type']).agg(
    Number_Of_Releases = ('show_id', 'count')
).reset_index()
```

```python
number_of_releases_by_release_year
```

```python
plt.figure(figsize=(12, 3))
plt.title("Number of Releases By Released Year")
plt.xlabel("Released Year")
plt.ylabel("Number of Releases")
sns.lineplot(data=number_of_releases_by_release_year, x='release_year', y='Number_Of_Releases', hue='type')
plt.show()
```

**Plot a Bar Plot to compare number of Movies and TV Shows released by last 10 years**

```python
number_of_releases_by_last_10_years = netflix_data[netflix_data['release_year'] >= max_release_year - 10].groupby(
    ['release_year', 'type']).agg(
        Number_Of_Releases = ('show_id', 'count')
).reset_index()
```

```python
number_of_releases_by_last_10_years
```

```python
plt.figure(figsize=(12, 3))
plt.title("Number of Released By Last 10 Years")
plt.xlabel("Released Year")
plt.ylabel("Number of Released")
sns.barplot(data=number_of_releases_by_last_10_years, x='release_year', y='Number_Of_Releases', hue='type')
plt.show()
```
